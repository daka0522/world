```python
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
WHITE = np.array([255, 255, 255])
GREEN = np.array([0, 255, 0])
BLACK = np.array([0, 0, 0])
```

```python
steps = 100
gradient = np.linspace(WHITE, GREEN, steps)
gradient2 = np.linspace(GREEN, WHITE, steps)
```

```python
full_gradient = np.concatenate((gradient, gradient2))
```

```python
# Visualize the gradient
plt.figure(figsize=(11, 3))
for i, color in enumerate(full_gradient):
    plt.axvline(x=i, color=color / 255, linewidth=4)
plt.axis('off')
plt.show()
```
